```python
def cal_fib_back(n):
    f0 = 0
    f1 = 1
    fn = None
    
    for i in range(n, 0, -1):
        if i == 0:
            return 0
        if i == 1:
            return 1
        fn = f0 + f1
        f1 = fn
        f0 = f1 
    return fn

print(cal_fib_back(5))
```
